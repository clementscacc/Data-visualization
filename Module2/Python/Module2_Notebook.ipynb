## 2. Some data

I put the match results in a pandas DataFrame so it is easy to work with.

```python
import pandas as pd

data = {
    "Match": ["M1", "M2", "M3", "M4", "M5", "M6", "M7", "M8"],
    "Goals For": [2, 1, 3, 0, 2, 4, 1, 2],
    "Goals Against": [1, 1, 2, 1, 0, 2, 2, 1],
}
df = pd.DataFrame(data)
df["Goal Diff"] = df["Goals For"] - df["Goals Against"]
df
```

```python
print("Total goals scored:", df["Goals For"].sum())
print("Average goals per match:", round(df["Goals For"].mean(), 2))
print("Wins:", (df["Goal Diff"] > 0).sum())
```

## 3. Graphic

A bar chart comparing goals for and goals against in each match.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(df))
width = 0.38

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(x - width/2, df["Goals For"], width, label="Goals For", color="#2a6fdb")
ax.bar(x + width/2, df["Goals Against"], width, label="Goals Against", color="#e8833a")
ax.set_xticks(x)
ax.set_xticklabels(df["Match"])
ax.set_ylabel("Goals")
ax.set_title("Goals For vs Goals Against by Match")
ax.legend()
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```
